```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
```

```python
ohe = OneHotEncoder(drop = 'first', sparse_output=False) #из n значений делаем n-1 признак, чтобы не было проблемы мультиколлиеарности
cat_encoded = ohe.fit_transform(data_train[cat_cols])
cat_encoded = pd.DataFrame(cat_encoded, columns = ohe.get_feature_names_out())
```

```python
data_train.reset_index(drop=True, inplace = True) #обнуляем индексы в исходном датафрейме, чтобы не было проблем
#при объединении его с encoded cat_cols, тк в data я удалял строчки с Nanами в TotalSpent
```

```python
data_train = pd.concat([data_train, cat_encoded], axis = 1)
data_train = data_train.drop(columns = cat_cols, axis = 1)
```

```python
y = data_train[target_col]
data_train = data_train.drop(columns = target_col, axis = 1)
data_train.head()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold


pipe_Log = Pipeline(steps = [('StSc', StandardScaler()), ('LogReg', LogisticRegression())])
kfolds = StratifiedKFold(n_splits=5, shuffle = True, random_state = 52) #нужно, чтобы модель перемешивала данные, по дефолту cv=5 в GridSearch не будет их мешать

param = {'LogReg__C': [100, 10, 1, 0.1, 0.01, 0.001]} #GridSearch не понимает, к чему мы обращаемся в пайплане,
#поэтому надо сначала указать название элемента в пайплайне, а затем двойное подчеркивание и название параметра
CvModel = GridSearchCV(estimator = pipe_Log,
                       param_grid=param,
                       cv = kfolds,
                       scoring = 'roc_auc',
                       n_jobs=-1,
                       refit=True) #тк мы поставили refit=true, то можем обращаться для предсказания к самой CvModel
CvModel.fit(data_train, y)
print(CvModel.best_params_)
print(CvModel.best_score_)
```

```python
import catboost


parameter = {'l2_leaf_reg': [0.5, 1, 3], 'depth': [4, 6, 8], 'learning_rate': [0.03, 0.05, 0.1]}

boosting_model = catboost.CatBoostClassifier(n_estimators=200) #вторая часть после AUC нужна чтобы он не высвечивал постоянно Metric AUC is not calculated on train by default. To calculate this metric on train, add hints=skip_train~false to metric parameters.

BoostModel = GridSearchCV(estimator = boosting_model,
                       param_grid= parameter,
                       cv = kfolds,
                       scoring = 'roc_auc',
                       n_jobs=-1,
                       refit=True)

BoostModel.fit(data_train_copy, y, cat_features=cat_cols)
print(BoostModel.best_params_)
print(BoostModel.best_score_)



#Ошибка возникает из-за того, как GridSearchCV (из Scikit-learn) пытается клонировать (скопировать)
#переданный ему оценщик (estimator) для каждой комбинации гиперпараметров. Этот процесс клонирования
#требует, чтобы все параметры, переданные в конструктор, были неизменяемыми и могли быть корректно скопированы.

# Почему cat_features вызывает проблему?

# Внутреннее устройство CatBoost: Параметр cat_features — это не просто обычный параметр, который передаётся
#в конструктор и хранится как есть. CatBoost внутри себя может модифицировать, проверять или по-особенному
#хранить этот параметр (например, преобразовывать список в tuple, проверять индексы и т.д.).
# Клонирование в Scikit-learn: Когда GridSearchCV клонирует модель, он проверяет, совпадают ли значения всех
#параметров, переданных в конструктор, с теми, что хранятся в объекте после инициализации. Если они не совпадают
#(а в случае cat_features могут быть незначительные изменения, например, список преобразован в кортеж), возникает ошибка RuntimeError.
# Обходной путь: Передача cat_features в fit(), а не в конструктор, позволяет избежать этой проблемы, потому что:

# GridSearchCV клонирует модель без указания cat_features.
# При вызове fit() для каждой клонированной модели, GridSearchCV передаёт cat_features через аргументы метода fit. Это поддерживается CatBoost.
# Параметр cat_features не становится частью состояния, которое проверяется при клонировании.
```

# Интерпретация модели: что на самом деле гонит отток

ROC-AUC 0.847 говорит, что модель ранжирует клиентов. Но бизнесу нужен ответ на другой вопрос:
**какие клиенты уходят и что с этим делать**. Ниже — разбор коэффициентов, проверка их на сырых
данных и оценка того, насколько модель полезна для таргетинга удержания.

```python
import numpy as np

# Достаём коэффициенты логистической регрессии.
# Признаки стандартизованы, поэтому коэффициенты сопоставимы между собой.
# exp(coef) = odds ratio: во сколько раз меняются шансы оттока при росте признака на 1 ст. отклонение.
best_lr = CvModel.best_estimator_.named_steps['LogReg']

coefs = (pd.DataFrame({
            'Признак': data_train.columns,
            'Коэффициент': best_lr.coef_[0],
            'Odds ratio': np.exp(best_lr.coef_[0])
         })
         .sort_values('Коэффициент'))

print('=== СНИЖАЮТ вероятность оттока ===')
display(coefs.head(8).round(3))
print('=== ПОВЫШАЮТ вероятность оттока ===')
display(coefs.tail(8).iloc[::-1].round(3))
```

### Что показывают коэффициенты

**Главный фактор удержания — длительный контракт.** Двухлетний контракт снижает шансы оттока
более чем вдвое (odds ratio 0.46), годовой — на четверть (0.76). Базой сравнения выступает
помесячный контракт, то есть именно он и есть зона риска.

**Главный фактор оттока — оптоволоконный интернет** (odds ratio 2.09, шансы вырастают вдвое).
Это классическая для телекома картина: оптика дороже, ожидания клиента выше, и любой сбой
переживается острее. Признак стоит рассматривать не как «оптика плохая», а как маркер
чувствительного к цене и качеству сегмента.

Дальше по убыванию — дополнительные подписки (фильмы, ТВ), безбумажный биллинг и оплата
электронным чеком. Все они, скорее всего, маркируют один и тот же тип клиента: технологичного,
привыкшего управлять подпиской самостоятельно, а значит легко её и отменяющего.

### Внимание: у срока жизни клиента коэффициент положительный — и это не ошибка

`ClientPeriod` получил коэффициент **+0.273**, то есть формально «чем дольше клиент с нами,
тем выше риск оттока». Это противоречит здравому смыслу, и ниже мы увидим, что в сырых данных
всё ровно наоборот.

Причина — **мультиколлинеарность с `TotalSpent`**. Суммарные траты — это почти произведение
срока жизни на месячный чек, поэтому два признака несут пересекающуюся информацию.
`TotalSpent` забрал на себя эффект лояльности (у него самый сильный отрицательный
коэффициент −1.204), а `ClientPeriod` в такой связке получил остаточный знак.

**Практический вывод:** коэффициенты коррелирующих признаков нельзя читать по отдельности.

# Насколько модель полезна для удержания

ROC-AUC — метрика для отчёта. Для бизнеса важнее другое: если у нас есть бюджет обзвонить
только часть базы, **сколько реально уходящих клиентов мы поймаем**.

Считаем честно — по предсказаниям на отложенных фолдах (out-of-fold), чтобы модель
не оценивала данные, на которых обучалась.

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import roc_auc_score

# Предсказания на отложенных фолдах: каждый клиент оценён моделью, которая его не видела.
proba_oof = cross_val_predict(CvModel.best_estimator_, data_train, y,
                              cv=kfolds, method='predict_proba')[:, 1]
print('Out-of-fold ROC-AUC:', round(roc_auc_score(y, proba_oof), 4))

base_rate = y.mean()
rows = []
for k in [0.05, 0.10, 0.15, 0.20, 0.30]:
    n_top = int(len(y) * k)
    top_idx = np.argsort(-proba_oof)[:n_top]
    hits = y.iloc[top_idx].sum()
    precision = hits / n_top
    rows.append({
        'Охват базы': f'{k:.0%}',
        'Клиентов': n_top,
        'Из них ушли': int(hits),
        'Precision': f'{precision:.1%}',
        'Recall': f'{hits / y.sum():.1%}',
        'Lift': f'{precision / base_rate:.2f}x'
    })

print(f'Базовая доля оттока в выборке: {base_rate:.1%}')
display(pd.DataFrame(rows))
```

### Результат таргетинга

| Охват базы | Клиентов | Из них ушли | Precision | Recall | Lift |
|---|---|---|---|---|---|
| 5% | 264 | 226 | **85.6%** | 16.3% | 3.27x |
| 10% | 528 | 401 | **75.9%** | 29.0% | 2.90x |
| 15% | 792 | 558 | 70.5% | 40.3% | 2.69x |
| 20% | 1 056 | 712 | 67.4% | 51.4% | 2.57x |
| 30% | 1 584 | 931 | 58.8% | 67.3% | 2.24x |

Если обзванивать **10% базы** с наибольшим риском, то трое из четырёх в этом списке
действительно собирались уйти — против одного из четырёх при случайном обзвоне.
Это **втрое эффективнее случайного выбора**, и за один заход мы накрываем 29% всех уходящих.

Расширение охвата до 30% ловит две трети оттока, но точность падает до 58.8% — почти
половина звонков уходит тем, кто и так остался бы. Где именно проводить границу,
зависит от экономики: стоимости контакта, размера удерживающей скидки и LTV клиента.
При дорогом удержании (персональный менеджер, крупная скидка) выгоднее узкий охват 5–10%,
при дешёвом (пуш, автообзвон) — широкий.
